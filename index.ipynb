```python
#independent and dependent feattures

X = pd.DataFrame(boston.data, columns=boston.feature_names)

# y is pulled directly from the target array
y = boston.target
```

```python
print("X shape:", X.shape)  # Expected: (20640, 8)
print("y shape:", y.shape) 
```

```python
X.head()
```

```python
#train test split

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
X_train
```

```python
#strndardizing the dataset
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
```

```python
X_train
```

```python
X_test=scaler.transform(X_test)
```

```python
X_test
```

```python
y_train.shape
```

```python
from sklearn.linear_model import LinearRegression

regression=LinearRegression()
```

```python
regression.fit(X_train,y_train)
```

```python
print(regression.coef_)
```

```python
print(regression.intercept_)
```

```python
#on which parameters model will be trained
regression.get_params()
```

```python
#prediction with the test data
reg_pred=regression.predict(X_test)
```

```python
reg_pred
```

```python
##plot a scatter plot for the prediction
plt.scatter(y_test,reg_pred)
```

```python
#reduals
residuals=y_test-reg_pred
residuals
```

```python
#plot with residuals

sns.displot(residuals,kind="kde")
```

```python
#scatter plot with predictions and residuals
plt.scatter(reg_pred,residuals)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error

print(mean_absolute_error(y_test,reg_pred))
print(mean_squared_error(y_test,reg_pred))
print(np.sqrt(mean_squared_error(y_test,reg_pred)))
```

```python
#adjusted  R square and R square
#r^2=1-ssr/sst
from sklearn.metrics import r2_score
score=r2_score(y_test,reg_pred)
print(score)
```

```python
#adjusted r2=1-[(1-r2)*(n-1)/(n-k-1)]
1-(1-score)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```

```python
#ajusted R2 is less than r2
```
